Extracting data files from .h5 format

```python
import pandas as pd
```

```python
hdf_file = '../../Dataset/UK-DALE/ukdale.h5'
```

```python
# Check available keys
with pd.HDFStore(hdf_file, 'r') as store:
    keys = store.keys()
    print(len(keys))

    for index,key in enumerate(keys):
        df = store[key]
        csv_name = "data/UK-DALE/"+key.strip('/').replace('/','_')+".csv"
        df.to_csv(csv_name)
        print(f"Saved {index + 1}st saved as {csv_name}")
```
